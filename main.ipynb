```python

import os
import json
import requests

from bs4 import BeautifulSoup
from dotenv import load_dotenv
from google import genai
from google.genai import types
from IPython.display import Markdown, display

load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY")

if not api_key:
    raise ValueError("GEMINI_API_KEY bulunamadı!")

gemini = genai.Client(api_key=api_key)

MODEL = "gemini-3.5-flash-lite"

print("Kütüphaneler yüklendi, Gemini istemcisi hazır!")
```

```python

response = gemini.models.generate_content(
    model=MODEL,
    contents="Merhaba! AI Job Analyzer projesi için tek cümlelik bir karşılama mesajı yaz."
)

print(response.text)
```

```python

import requests

url = "https://www.python.org/jobs/"

response = requests.get(
    url,
    headers={"User-Agent": "AI-Job-Analyzer/1.0"},
    timeout=15
)

print("HTTP Status Code:", response.status_code)
print("HTML uzunluğu:", len(response.text))
```

```python

from bs4 import BeautifulSoup

response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

for element in soup(["script", "style", "nav", "footer", "header"]):
    element.decompose()

text = soup.get_text(separator="\n", strip=True)

print(text[:2000])
```

```python

def fetch_job_content(url):
    headers = {
        "User-Agent": "AI-Job-Analyzer/1.0"
    }

    response = requests.get(
        url,
        headers=headers,
        timeout=15
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    for element in soup(["script", "style", "nav", "footer", "header"]):
        element.decompose()

    content = soup.get_text(separator="\n", strip=True)

    return content
```

```python
job_text = fetch_job_content("https://www.python.org/jobs/")

print(job_text[:3000])
```

```python

from urllib.parse import urljoin

job_links = []

for link in soup.find_all("a", href=True):
    href = link["href"]

    if "/jobs/" in href:
        full_url = urljoin(url, href)
        job_links.append(full_url)

job_links = list(dict.fromkeys(job_links))

print(job_links[:10])
```

```python
for index, link in enumerate(job_links[:10]):
    print(index, ":", link)
```

```python
job_url = job_links[0]

job_text = fetch_job_content(job_url)

print(job_text[:2000])
```

```python

job_system_prompt = """
You are an expert AI job posting analyzer.

Your task is to analyze job postings and extract important information.

Identify:
- Job title
- Company name
- Required technical skills
- Experience requirements
- Job responsibilities

Rules:
- Only use information provided in the job posting.
- Do not invent missing information.
- Respond in valid JSON format.
"""
```

```python

def get_job_user_prompt(job_text):
    user_prompt = f"""
    Analyze the following job posting.

    Extract the job title, company name,
    required technical skills, experience requirements,
    and responsibilities.

    Job Posting:
    {job_text}
    """

    return user_prompt
```

```python
job_user_prompt = get_job_user_prompt(job_text)

print(job_user_prompt[:1000])
```

```python
response = gemini.models.generate_content(
    model=MODEL,
    contents=get_job_user_prompt(job_text),
    config=types.GenerateContentConfig(
        system_instruction=job_system_prompt,
        response_mime_type="application/json"
    )
)

analysis = json.loads(response.text)

print(json.dumps(analysis, indent=4, ensure_ascii=False))
```

```python

def analyze_job(job_text):

    response = gemini.models.generate_content(
        model=MODEL,
        contents=get_job_user_prompt(job_text),
        config=types.GenerateContentConfig(
            system_instruction=job_system_prompt,
            response_mime_type="application/json"
        )
    )

    analysis = json.loads(response.text)

    return analysis
```

```python
job_analysis = analyze_job(job_text)

print(json.dumps(job_analysis, indent=4, ensure_ascii=False))
```

```python

report_system_prompt = """
You are a professional career advisor.

Your task is to create a clear and well-structured
job analysis report based on the provided JSON data.

The report must include:
1. Job title and company name
2. Short summary of the position
3. Required technical skills
4. Experience requirements
5. Main responsibilities
6. Preparation recommendations

Rules:
- Write the report in Turkish.
- Use Markdown formatting.
- Use clear headings and bullet points.
- Do not invent job requirements.
- If information is missing, say it is not specified.
- Clearly separate your recommendations from the job requirements.
"""
```

```python

def get_report_user_prompt(job_analysis):

    analysis_json = json.dumps(
        job_analysis,
        indent=2,
        ensure_ascii=False
    )

    user_prompt = f"""
    Create a professional job analysis report
    using the following structured information.

    Job Analysis:
    {analysis_json}

    Write the report in Turkish using Markdown.
    """

    return user_prompt
```

```python
report_user_prompt = get_report_user_prompt(job_analysis)

print(report_user_prompt[:1500])
```

```python

response = gemini.models.generate_content(
    model=MODEL,
    contents=get_report_user_prompt(job_analysis),
    config=types.GenerateContentConfig(
        system_instruction=report_system_prompt
    )
)

report = response.text

print(report)
```
